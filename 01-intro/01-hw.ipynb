### Q7. Sum of weights

```python
X = df[df['origin'] == 'Asia'].reset_index().loc[:6,['vehicle_weight','model_year']].to_numpy()
X
```

```python
XT = X.T
XT
```

```python
XTX = XT.dot(X)
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv.dot(XT).dot(y)
w
```

```python
w.sum().round(4)
```
